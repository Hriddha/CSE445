```python
import os
import cv2
import numpy as np

# Directory containing digit folders 0 through 9
data_dir = "./"  # Update this path if the folders are in a subfolder

data = []
labels = []

# Loop through each digit folder (0 to 9)
for digit in range(10):
    folder_path = os.path.join(data_dir, str(digit))
    
    if not os.path.exists(folder_path):
        continue

    for img_name in os.listdir(folder_path):
        img_path = os.path.join(folder_path, img_name)
        
        # Read image in grayscale
        img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
        
        if img is not None:
            # Resize image to a fixed size (e.g., 28x28 like MNIST) if needed
            img = cv2.resize(img, (28, 28))
            
            data.append(img)
            labels.append(digit)

# Convert lists to NumPy arrays
data_array = np.array(data)
labels_array = np.array(labels)


np.save("dataset_2.npy", data_array)
np.save("labels_2.npy", labels_array)

print(f"Data saved successfully!")
print(f"Dataset shape: {data_array.shape}")
print(f"Labels shape: {labels_array.shape}")
```
